```python
import os
from rdkit import Chem
from rdkit.Chem import Draw
from rdkit.Chem.Draw import rdMolDraw2D
from IPython.display import display, Image

# 1. Folder setup jahan saari High-Res images save hongi
output_dir = "HighRes_Molecule_Images"
os.makedirs(output_dir, exist_ok=True)

# 2. Tera complete NXI series dictionary (Baseline se lekar Rigid/Indane tak)
# Agar koi aur naya molecule add karna ho, toh bas yahan uska naam aur SMILES daal dena
smiles_dict = {
    "NXI-07_Baseline_Clean": "CC1CCC(NC(=O)Nc2ccc(CCC(=O)CO)cc2)CC1",
    "NXI-09_Gold_Tier": "O=C(CO)CCc1ccc(NC(=O)N[C@]2(C3)C[C@]4(C)C[C@H]3C4(C)C2)cc1", # Added based on your ketol+adamantane design
    "NXI-12_Rigid_Alkene": "O=C(/C=C/c1ccc(NC(=O)N[C@]2(C3)C[C@]4(C)C[C@H]3C4(C)C2)cc1)CO",
    "NXI-13_Rigid_Lead": "O=C(Cc1ccc(NC(=O)N[C@]2(C3)C[C@]4(C)C[C@H]3C4(C)C2)cc1)CO",
    "NXI-14A_Direct_Linker": "O=C(CO)c1ccc(NC(=O)N[C@]2(C3)C[C@]4(C)C[C@H]3C4(C)C2)cc1",
    "NXI-14B_Indane_Bulky": "O=C(CO)C1Cc2ccc(NC(=O)N[C@]3(C4)C[C@]5(C)C[C@H]4C5(C)C3)cc2C1"
}

mols = []
legends = []

print(f"🖼️ Generating High-Resolution 2D Structures...\n" + "-"*50)

# 3. Har molecule ko process karna aur High-Res PNG save karna
for name, smiles in smiles_dict.items():
    mol = Chem.MolFromSmiles(smiles)
    if mol is not None:
        # Standardize representation for better rendering
        Chem.rdCoordGen.AddCoords(mol)
        
        # High-Quality Drawing Setup (1000x1000 pixels resolution)
        drawer = rdMolDraw2D.MolDraw2DCairo(1000, 1000)
        
        # Thoda professional look ke liye drawing options tweak kar rahe hain
        opts = drawer.drawOptions()
        opts.bondLineWidth = 3
        opts.scalingFactor = 40
        opts.clearBackground = True
        
        # Image draw aur save karna
        drawer.DrawMolecule(mol)
        drawer.FinishDrawing()
        
        file_path = os.path.join(output_dir, f"{name}.png")
        with open(file_path, 'wb') as f:
            f.write(drawer.GetDrawingText())
            
        print(f"✅ Saved: {name}.png (Resolution: 1000x1000)")
        
        # Jupyter grid ke liye lists mein add karna
        mols.append(mol)
        legends.append(name)
    else:
        print(f"❌ Error: Invalid SMILES for {name}")

# 4. Jupyter Notebook mein ek sath sabko Grid format mein display karna
print("\n📊 Molecule Grid Preview:")
grid_img = Draw.MolsToGridImage(
    mols, 
    molsPerRow=3, 
    subImgSize=(400, 400), 
    legends=legends, 
    returnPNG=True
)
display(grid_img)
```

```python
from rdkit import Chem
from rdkit.Chem import Draw, Descriptors, Crippen, Lipinski, rdDepictor
from rdkit.Chem.Draw import rdMolDraw2D
from IPython.display import Image, display
import os

# High-quality 2D coordinate generation (rings ekdum seedhi aur saaf banengi)
rdDepictor.SetPreferCoordGen(True)

# Stage 3 (Gen-2 Clean/Gold) & Stage 4 (Rigid Champions) Exact SMILES Dictionary
molecules_data = {
    # --- STAGE 3: GEN-2 CLEAN & GOLD BREAKTHROUGH ---
    "01_Stage3_NXI_09_Gold_Ketol": {
        "title": "NXI-09-Gold (Bornyl + Carbon Linker + Ketol ZBG)",
        "smiles": "CC1(C)C2CCC1(C)C(NC(=O)Nc3ccc(CCCCC(=O)CO)cc3)C2",
        "note": "Stage 3: Non-Mutagenic Ketol (-COCH2OH) | Bornyl Cage"
    },
    "02_Stage3_NXI_10_Pyridylamide": {
        "title": "NXI-10 (Bornyl + Carbon Linker + 2-Pyridylamide ZBG)",
        "smiles": "CC1(C)C2CCC1(C)C(NC(=O)Nc3ccc(CCCCC(=O)Nc4ccccn4)cc3)C2",
        "note": "Stage 3: Bidentate 2-Aminopyridine ZBG"
    },
    "03_Stage3_NXI_11_N_Methyl_Hydroxamate": {
        "title": "NXI-11 (Bornyl + Sterically Shielded N-Me Hydroxamate)",
        "smiles": "CC1(C)C2CCC1(C)C(NC(=O)Nc3ccc(CCCCC(=O)N(C)O)cc3)C2",
        "note": "Stage 3: N-Methyl blocks Lossen Rearrangement"
    },

    # --- STAGE 4: THE RIGID CHAMPIONS (ROTATABLE BOND LOCKS) ---
    "04_Stage4_NXI_12_Cinnamoyl_Lock": {
        "title": "NXI-12 (Cinnamoyl Trans-Alkene Rigid Lock)",
        "smiles": "CC1(C)C2CCC1(C)C(NC(=O)Nc3ccc(/C=C/CCC(=O)CO)cc3)C2",
        "note": "Stage 4: Trans-Alkene (-CH=CH-) locks linker geometry"
    },
    "05_Stage4_NXI_13_Indane_Rigid_Lead": {
        "title": "NXI-13 (Indane Fused-Ring Lock - MAIN RIGID LEAD)",
        "smiles": "CC1(C)C2CCC1(C)C(NC(=O)Nc3ccc4CCC(CC(=O)CO)c4c3)C2",
        "note": "Stage 4: Fused Indane Ring eliminates chain flop"
    },
    "06_Stage4_NXI_14_BCP_Indane_UltraCompact": {
        "title": "NXI-14 (BCP Cage + Indane Ring + Ketol ZBG)",
        "smiles": "CC12CC(C1)(C2)NC(=O)Nc3ccc4CCC(CC(=O)CO)c4c3",
        "note": "Stage 4: [1.1.1]Bicyclopentane (BCP) Cage | 3 Rot. Bonds"
    }
}

output_dir = "NXI_Stage3_Stage4_HD_Images"
os.makedirs(output_dir, exist_ok=True)

mols_for_grid = []
legends_for_grid = []

print("="*75)
print("  GENERATING ULTRA-HD 2D STRUCTURES FOR HAND-DRAWING & VISUALIZATION")
print("="*75)

for filename, info in molecules_data.items():
    mol = Chem.MolFromSmiles(info["smiles"])
    if mol is None:
        print(f"Error parsing SMILES for {filename}")
        continue
        
    rdDepictor.Compute2DCoords(mol)
    
    # Real-time RDKit Calculations (Zero Contradiction!)
    mw = Descriptors.MolWt(mol)
    tpsa = Descriptors.TPSA(mol)
    clogp = Crippen.MolLogP(mol)
    rot_bonds = Lipinski.NumRotatableBonds(mol)
    hbd = Lipinski.NumHDonors(mol)
    hba = Lipinski.NumHAcceptors(mol)
    
    legend_text = (
        f"{info['title']}\n"
        f"{info['note']}\n"
        f"MW: {mw:.1f} | TPSA: {tpsa:.2f} A^2 | cLogP: {clogp:.2f} | "
        f"RotBonds: {rot_bonds} | HBD: {hbd}"
    )
    
    # High-Resolution Cairo PNG Drawer (1200 x 850 px)
    drawer = rdMolDraw2D.MolDraw2DCairo(1200, 850)
    opts = drawer.drawOptions()
    opts.bondLineWidth = 3.0              # Moti, saaf lines taaki copy pe draw karna aasan ho
    opts.scaleBondWidth = True
    opts.legendFontSize = 22              # Bada, clear text
    opts.padding = 0.08
    opts.addStereoAnnotation = True
    
    drawer.DrawMolecule(mol, legend=legend_text)
    drawer.FinishDrawing()
    
    img_path = os.path.join(output_dir, f"{filename}.png")
    with open(img_path, "wb") as f:
        f.write(drawer.GetDrawingText())
        
    print(f"\n✅ {info['title']}")
    print(f"   SMILES   : {info['smiles']}")
    print(f"   Stats    : MW={mw:.2f} | TPSA={tpsa:.2f} Å² | cLogP={clogp:.2f} | RotBonds={rot_bonds}")
    print(f"   Saved to : {img_path}")
    
    # Display directly inside Jupyter Notebook
    display(Image(filename=img_path, width=750))
    
    mols_for_grid.append(mol)
    legends_for_grid.append(f"{info['title'].split('(')[0].strip()}\nTPSA: {tpsa:.1f} | cLogP: {clogp:.2f} | RB: {rot_bonds}")

# Ek Master Comparison Sheet (All 6 in 1 Image) bhi save kar le
grid_img = Draw.MolsToGridImage(
    mols_for_grid,
    molsPerRow=2,
    subImgSize=(700, 500),
    legends=legends_for_grid,
    returnPNG=True
)

grid_path = os.path.join(output_dir, "00_MASTER_GRID_Stage3_and_Stage4.png")
with open(grid_path, "wb") as f:
    f.write(grid_img.data if hasattr(grid_img, 'data') else grid_img)

print("\n" + "="*75)
print(f"🎯 MASTER GRID SAVED AT: {grid_path}")
print("="*75)
```
